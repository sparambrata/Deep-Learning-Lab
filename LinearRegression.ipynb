```python
#23070521101 Parambrata Sanyal

import numpy as np

# -----------------------------
# Step 1: Define the Dataset
# -----------------------------
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([20, 40, 50, 65, 80], dtype=float)

print("="*60)
print("Original Dataset")
print("="*60)
print("X:", x)
print("Y:", y)

# -----------------------------
# Step 2: Calculate Means
# -----------------------------
x_mean = np.mean(x)
y_mean = np.mean(y)

print("\nMean of X =", x_mean)
print("Mean of Y =", y_mean)

# -----------------------------
# Step 3: Calculate Slope
# Formula:
# m = Σ[(xi-x̄)(yi-ȳ)] / Σ[(xi-x̄)^2]
# -----------------------------

numerator = np.sum((x - x_mean) * (y - y_mean))
denominator = np.sum((x - x_mean) ** 2)

slope = numerator / denominator

print("\nSlope Calculation")
print("----------------------")
print("Numerator =", numerator)
print("Denominator =", denominator)
print("Slope (m) =", slope)

# -----------------------------
# Step 4: Calculate Intercept
# Formula:
# c = ȳ - m*x̄
# -----------------------------

intercept = y_mean - slope * x_mean

print("\nIntercept (c) =", intercept)

# -----------------------------
# Step 5: Regression Equation
# -----------------------------

print("\nRegression Equation")
print("----------------------")
print(f"y = {slope:.2f}x + {intercept:.2f}")

# -----------------------------
# Step 6: Predict Y
# -----------------------------

y_pred = slope * x + intercept

print("\nPredicted Values")
print("----------------------")

for i in range(len(x)):
    print(f"x = {x[i]:.0f}   Predicted y = {y_pred[i]:.2f}")

# -----------------------------
# Step 7: Residual Error
# Formula:
# Residual = Actual - Predicted
# -----------------------------

residuals = y - y_pred

print("\nResidual Errors")
print("----------------------")

for i in range(len(x)):
    print(f"Actual = {y[i]:.0f}   Predicted = {y_pred[i]:.2f}   Residual = {residuals[i]:.2f}")

# -----------------------------
# Step 8: Mean Absolute Error
# -----------------------------

absolute_error = np.abs(residuals)

MAE = np.mean(absolute_error)

print("\nAbsolute Errors =", absolute_error)
print("Mean Absolute Error (MAE) =", MAE)

# -----------------------------
# Step 9: Mean Squared Error
# -----------------------------

squared_error = residuals ** 2

MSE = np.mean(squared_error)

print("\nSquared Errors =", squared_error)
print("Mean Squared Error (MSE) =", MSE)

# -----------------------------
# Step 10: Root Mean Squared Error
# -----------------------------

RMSE = np.sqrt(MSE)

print("\nRoot Mean Squared Error (RMSE) =", RMSE)

# -----------------------------
# Step 11: Final Summary
# -----------------------------

print("\n" + "="*60)
print("FINAL SUMMARY")
print("="*60)

print("Mean of X               :", x_mean)
print("Mean of Y               :", y_mean)
print("Slope                   :", slope)
print("Intercept               :", intercept)
print("Regression Equation     : y = {:.2f}x + {:.2f}".format(slope, intercept))
print("MAE                     :", MAE)
print("MSE                     :", MSE)
print("RMSE                    :", RMSE)

print("="*60)
```
